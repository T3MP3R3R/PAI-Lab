```python
import cv2
from google.colab.patches import cv2_imshow

image_path = "example.jpg"
img = cv2.imread(image_path)

if img is None:
    print("Error: Image not found or unable to load")
    exit()

cv2_imshow(img)
cv2.waitKey(0)

gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_img)
cv2.waitKey(0)

new_width = 300
new_height = 200
resized_img = cv2.resize(img, (new_width, new_height))
cv2_imshow(resized_img)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

**Task 5:**

```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np

image_path = "example.jpg"  # replace with your file
img_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

cv2_imshow(img_gray)

thresh_value = 127
max_value = 255
_, binary_img = cv2.threshold(img_gray, thresh_value, max_value, cv2.THRESH_BINARY)

cv2_imshow(binary_img)

(h, w) = binary_img.shape
center = (w//2, h//2)

angle = 60
scale = 1.0
M = cv2.getRotationMatrix2D(center, angle, scale)

rotated_img = cv2.warpAffine(binary_img, M, (w, h))

cv2_imshow(rotated_img)
```

```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np

img1 = cv2.imread("img1.jpg")
img2 = cv2.imread("img2.jpg")

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

alpha = 0.6
beta = 0.4
blended = cv2.addWeighted(img1, alpha, img2, beta, 0)

cv2_imshow(blended)

gray = cv2.cvtColor(blended, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)

equalized = cv2.equalizeHist(gray)
cv2_imshow(equalized)
```
